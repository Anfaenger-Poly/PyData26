### 기본 통계 함수

```python
import pandas as pd
titanic_data = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
titanic_df = pd.read_csv(titanic_data, sep="\t")
titanic_df.info()
```

```python
# 컬럼 소문자
new_columns = []
for col in titanic_df.columns:
    # new_columns.append(col.lower())
    new_columns.append(col[0].lower() + col[1:])
titanic_df.columns = new_columns
```

```python
# 데이터의 통계치 확인하기
titanic_df.describe()
```

```python
# 합계 구하기 : sum 함수
titanic_df.sum()
```

```python
# 총 개수 구하기 : count 함수
titanic_df.count()
```

```python
# 최댓값 구하기 : max 함수
titanic_df.max()
```

```python
# 최솟값 구하기 : min 함수
titanic_df.min()
```

```python
# 평균값 구하기 : mean 함수, 버전 3.10은 (numeric_only=True) 없이
titanic_df.mean(numeric_only=True)
```

```python
# 중간값 구하기 : median 함수
titanic_df.median(numeric_only=True)
```

```python
# 최빈값 구하기 : mad = mode 함수
titanic_df.mode(numeric_only=True)
```

```python
# 분산 구하기 : var 함수
titanic_df.var(numeric_only=True)
```
